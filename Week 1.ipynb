```python
df["Date"] = pd.to_datetime( df["Period"].astype(str),format="%Y.%m")
```

```python
df[["Period", "Date"]].head()
```

```python
df["Year"] = df["Date"].dt.year
df["Quarter"] = df["Date"].dt.quarter
```

```python
print("Start Date:", df["Date"].min())
print("End Date:", df["Date"].max())
```

```python
df["Group"].value_counts()
```

```python
jobs = df[(df["Series_title_1"] == "Filled jobs") &(df["Series_title_3"] == "Actual")].copy()
```

```python
jobs.shape
```

```python
total_jobs = jobs[jobs["Series_title_2"] == "Total Industry"].copy()
```

```python
total_jobs[["Date", "Data_value"]].head()
```

```python
total_jobs.plot(
    x="Date",
    y="Data_value",
    figsize=(12, 5),
    legend=False)
plt.title("Total Industry Filled Jobs Over Time")
plt.xlabel("Year")
plt.ylabel("Filled Jobs")
plt.grid(True)
plt.show()
```

```python
industry_jobs = jobs[(jobs["Group"] == "Industry by employment variable") &(jobs["Series_title_2"] != "Total Industry")].copy()
```

```python
latest_date = industry_jobs["Date"].max()

latest_industry = industry_jobs[
    industry_jobs["Date"] == latest_date
].sort_values(
    "Data_value",
    ascending=False
)

latest_industry[
    ["Series_title_2", "Data_value"]
].head(10)
```

```python
top10 = latest_industry.head(10)
plt.figure(figsize=(12, 6))
sns.barplot(
    data=top10,
    x="Data_value",
    y="Series_title_2")
plt.title("Top 10 Industries by Filled Jobs")
plt.xlabel("Filled Jobs")
plt.ylabel("Industry")
plt.show()
```

```python
region_jobs = jobs[jobs["Group"] == "Region by employment variable"].copy()
```

```python
latest_region = region_jobs[
    region_jobs["Date"] == latest_date
].sort_values(
    "Data_value",
    ascending=False
)

latest_region[
    ["Series_title_2", "Data_value"]
].head(10)
```

```python
top_regions = latest_region.head(10)
age_jobs = jobs[
    jobs["Group"] == "Age by employment variable"
].copy()plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_regions,
    x="Data_value",
    y="Series_title_2")
plt.title("Top Regions by Filled Jobs")
plt.xlabel("Filled Jobs")
plt.ylabel("Region")
plt.show()
```

```python
age_jobs = jobs[jobs["Group"] == "Age by employment variable"].copy()
```

```python
latest_age = age_jobs[
    age_jobs["Date"] == latest_date
].sort_values(
    "Data_value",
    ascending=False
)

latest_age[
    ["Series_title_2", "Data_value"]
]
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(
    data=latest_age,
    x="Series_title_2",
    y="Data_value"
)
plt.title("Filled Jobs by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Filled Jobs")
plt.xticks(rotation=45)
plt.show()
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing.values,
    y=missing.index
)
plt.title("Missing Values by Column")
plt.xlabel("Number of Missing Values")
plt.ylabel("Column")
plt.show()
```
