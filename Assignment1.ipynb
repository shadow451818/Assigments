```python
Student_marks={
    "Student1":78,
    "Student2":45,
    "Student3":60,
    "Student4":79,
    "Student5":75,
    "Student6":55,
    "Student7":44,
    "Student8":48,
    "Student9":80,
    "Student10":91,
    "Student11":88,
    "Student12":28,
    "Student13":68,
    "Student14":38,
    "Student15":29,
    "Student16":31,
    "Student17":22,
    "Student18":54,
    "Student19":10,
    "Student20":19
} 
avg_marks = sum(Student_marks.values())/len(Student_marks)

print("Average marks:", avg_marks)

print("\nStudents who have marks above average:")

for student, marks in Student_marks.items():
    if marks > avg_marks:
        print(student, ":", marks)
```

```python
def assign_grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    else:
        return "F"


students_marks={
    "Rahul": 78,
    "Aman": 95,
    "Priya": 92,
    "Neha": 45,
    "Rohit": 66
}

for student,marks in students_marks.items():
    grade = assign_grade(marks)
    print(student,"- Marks:", marks, "- Grade:",grade)
```
